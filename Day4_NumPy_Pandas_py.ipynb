```python
import matplotlib.pyplot as plt
import numpy as np

#line chart
x=[1,2,3,4,5]
y=[20,40,60,80,100]
plt.plot(x, y, color='Red', marker='o')
plt.title('line chart')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
#bar chart
plt.bar(x,y,color='green')
plt.title('bar chart')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
#step 2
import seaborn as sns
import pandas as pd
#student data
data={
    "name":['kashif','yasir','khan','jan'],
    'Cgpa':[2,2.5,3,3.4],
    'Projects':[1,2,3,4]

}
df=pd.DataFrame(data)
#beautiful bar chart
sns.barplot(x='name',y='Cgpa',data=df,palette='Blues')
plt.title('Student Cgpa comparsion')
plt.show()
#scatter plot
sns.scatterplot(x="Cgpa", y="Projects", data=df, size="Projects")
plt.title("Cgpa vs Projects")
plt.show()
```
